```python
def f(x, k, b):
    return b*np.exp(x*k)

def dfdk(x, k, b):
    return b*np.exp(x*k)*x

def dfdb(x, k, b):
    return np.exp(x*k)
```

```python
X = np.random.randn(N)
Y = f(X, k_base, b_base) + spread*np.random.randn(N)

plt.scatter(X, Y)
plt.show()
```

```python
def E(k,b):
    return np.sum((f(X, k, b) - Y)**2)
```

```python
def gradE(k,b):
    return np.sum((f(X, k, b) - Y)*dfdk(X, k, b)), np.sum((f(X, k, b) - Y)*dfdb(X, k, b))
```

```python
k=1.0
b=1.0
print(E(k,b))
epochs=30
history = np.zeros(epochs)
for epoch in range(epochs):
    G=gradE(k,b)
    k = k - alpha*G[0]
    b = b - alpha*G[1]
    history[epoch] = E(k,b)
```

```python
plt.scatter(X, b*np.exp(k*X))
plt.show()
```
